## The key function for working with files in Python is the open() function.

1- The open() function takes two parameters :

        a- filename 
        b- mode

2 - There are four different methods (modes) for opening a file:

        "r" - Read - Default value. Opens a file for reading, error if the file does not exist

        "a" - Append - Opens a file for appending, creates the file if it does not exist

        "w" - Write - Opens a file for writing, creates the file if it does not exist

        "x" - Create - Creates the specified file, returns an error if the file exists

3- In addition you can specify if the file should be handled as binary or text mode

        "t" - Text - Default value. Text mode

        "b" - Binary - Binary mode (e.g. images)

```python
# Because "r" for read, and "t" for text are the default values, you do not need to specify them.
f = open("demofile.txt", "rt")
f
```

# Write to an Existing File

To write to an existing file, you must add a parameter to the open() function:

        "a" - Append - will append to the end of the file 

        "w" - Write - will overwrite any existing content
 
        if the file does not exist it creates a new file

## Append mode 

```python
# Open the file "demofile2.txt" and append content to the file:

f = open("demofile2.txt", "a")
f.write("Now the file has more content!")
f.close()
```

```python
with open('demofile2.txt','a') as f:
    f.write('This text has to be appended at the end')
```

```python
#open and read the file after the appending:
f = open("demofile2.txt", "r")
print(f.read())
f.close()
```

## Write mode

    Note: the "w" method will overwrite the entire file.

```python
with open('writing_file_example.txt','w') as f:
    f.write('hello new file text')
```

```python
f = open("writing_file_example.txt", "w")
f.write("Woops! I have deleted the content!")
f.close()
```

```python
#open and read the file after the overwriting:
f = open("writing_file_example.txt", "r")
print(f.read())
```

## Changing Dictionary to JSON

        To change a dictionary to a JSON we use dumps method from the json module.

```python
# python dictionary
person = {
    "name": "ahmed",
    "country": "Egypt",
    "city": "Cairo",
    "skills": ["JavaScrip", "React", "Python"]
}
```

```python
# let's convert it to  json
person_json = json.dumps(person, indent=4) # indent could be 2, 4, 8. It beautifies the json
print(type(person_json))
print(person_json)
```

```python
# python dictionary
person = {
    "name": "ahmed",
    "country": "Egypt",
    "city": "Cairo",
    "skills": ["JavaScrip", "React", "Python"]
}
with open('json_example.json', 'w', encoding='utf-8') as f:
    json.dump(person, f, ensure_ascii=False, indent=4)
```

2- CSV is a simple file format used to store tabular data, such as a spreadsheet or database.

```python
import csv
```

```python
with open('csv_example.csv') as f:
    csv_reader = csv.reader(f, delimiter=',') # we use reader method to read csv

print(csv_reader)
```

```python
csv_reader.line_num
```

```python
import csv

with open('csv_example.csv') as f:
    csv_reader = csv.reader(f, delimiter=',') # we use reader method to read csv
    line_count = 0
    for row in csv_reader:
        if line_count == 0:
            print(f'Column names are :{", ".join(row)}')
            line_count += 1
        else:
            print(
                f'\t{row[0]} is a developer. He lives in {row[1]}, {row[2]} and his skill is {row[3]}.')
            line_count += 1
    print(f'Number of lines:  {line_count}')
```

# files with xml extension

        XML is another structured data format which looks like HTML. 

        In XML the tags are not predefined. The first line is an XML declaration. 

        The person tag is the root of the XML. 

        The person has a gender attribute.

```python
# Example

# <?xml version="1.0"?>
# <person gender="male">   ---> Starting tag for person root element with gender attribute
#   <name>Mahmoud</name>   ---> This is a child element of person.
#   <country>Egypt</country> ---> This is a child element of person.
#   <city>Cairo</city>   ---> This is a child element of person.
#   <skills>            ---> Starting tag for skills which is a child element of person. which serves as a container for multiple skill 
#     <skill>C++</skill>  --> This is a child element of skills
#     <skill>React</skill>  --> This is a child element of skills
#     <skill>Python</skill>  --> This is a child element of skills
#   </skills>             ---> Closing tag for skills 
# </person>               ---> Closing tag for person
```

```python
import xml.etree.ElementTree as ET

# Parse XML file
tree = ET.parse('xml_example.xml')

# Get root element
root = tree.getroot()

# Print root tag, attribute, and child elements
print('Root tag:', root.tag)
print('Attribute:', root.attrib)

print('Child elements:')
for child in root:
    print('field: ', child.tag)
```
