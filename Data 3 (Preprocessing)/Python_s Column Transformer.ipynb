# 1- make column transformer

```python
ohe = OneHotEncoder(sparse_output=False)
```

```python
ode = OrdinalEncoder()
```

```python
ct = make_column_transformer(

    (ohe, ['State']),
    (ode, ['State']),
    remainder='passthrough'

    )

ct.set_output(transform='pandas')
```

```python
df_new = ct.fit_transform(df)
df_new.head()
```

# 3- one hot encoding

```python
df.State.unique()
```

```python
df.State.value_counts()
```

```python
encoder_hot = OneHotEncoder(handle_unknown= 'ignore',sparse_output= False).set_output(transform='pandas')

state_encoder_hot = encoder_hot.fit_transform(df[['State']])

state_encoder_hot
```

```python
df = pd.concat([df, state_encoder_hot], axis=1)
df.head()
```

```python
# Split features and target
X = df.drop("Profit", axis=1)
y = df["Profit"]

# Define categorical and numerical columns
categorical_cols = X.select_dtypes(include=['object']).columns
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns
```

```python
numerical_cols
```

```python
categorical_cols
```

```python
# Define preprocessing
preprocessor = make_column_transformer(
        (StandardScaler(), numerical_cols),
        (OneHotEncoder(drop='first' ,  sparse_output=False), categorical_cols)
    ).set_output(transform='pandas')
```

```python
X.shape, y.shape
```

```python
# Apply preprocessing
X_processed = preprocessor.fit_transform(X)
```

```python
X_processed.head()
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(X_processed, y, test_size=0.2, random_state=42)

print("Preprocessing complete. Shape of X_train:", X_train.shape)
```
