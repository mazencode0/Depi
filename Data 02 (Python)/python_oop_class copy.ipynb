Class Constructor

1- In the examples above, we have created an object from the Person class. However, a class without a constructor is not really useful in real applications. 

2- Let us use constructor function to make our class more useful. Like the constructor function in Java or JavaScript, Python has also a built-in init() constructor function. 

3- The init constructor function has self parameter which is a reference to the current instance of the class Examples:

```python
class Person:
      
      def __init__ (self, name , age): # (person_1 , 'Mahmoud' , 55)
        # self allows to attach parameter to the class
          self.name =name # person_1.name = 'Mahmoud'
          self.age = age  # person_1.age = 55
      
      def display(self,job ,country ) :
           
           return f"my name is {self.name} and my age is {self.age} and my job is {job} and I am from {country}"

      @staticmethod
      def greeting():
          print("hello world")

person_1 = Person('Mahmoud' , 55)
person_2    = Person('Ali' , 44)

print(person_1.display(job='data engineer' , country = 'egypt'))
```

Inheritance

1- Using inheritance we can reuse parent class code. 

2- Inheritance allows us to define a class that inherits all the methods and properties from parent class. 

3- The parent class or super or base class is the class which gives all the methods and properties. 

4- Child class is the class that inherits from another or parent class. 

Let us create a student class by inheriting from person class.

```python
# student class is the class that inherits from person class
class Student(Person):
    pass

s1 = Student('Ahmed', 'Ali', 30, 'Finland', 'Helsinki')
s2 = Student('Mostafa', 'Ibrahim', 28, 'Finland', 'Espoo')
print(s1.person_info())

s1.add_skill('JavaScript')
s1.add_skill('React')
s1.add_skill('Python')
print(s1.skills)

print(s2.person_info())
s2.add_skill('Organizing')
s2.add_skill('Marketing')
s2.add_skill('Digital Marketing')
print(s2.skills)
```

1- We did not call the init() constructor in the child class. If we didn't call it then we can still access all the properties from the parent. 

2- But if we do call the constructor we can access the parent properties by calling super.

3- We can add a new method to the child or we can override the parent class methods by creating the same method name in the child class. 

4- When we add the init() function, the child class will no longer inherit the parent's init() function.

```python
# Overriding parent method

class Student(Person):
    def __init__ (self, firstname='Ahmed', lastname='Ali',age=25, country='Finland', city='Helsinki', gender='male'):
        self.gender = gender
        super().__init__(firstname, lastname,age, country, city)
    def person_info(self):
        gender = 'He' if self.gender =='male' else 'She'
        return f'{self.firstname} {self.lastname} is {self.age} years old. {gender} lives in {self.city}, {self.country}.'
    


s1 = Student('Ahmed', 'Ali', 30, 'Finland', 'Helsinki','male')
s2 = Student('aya', 'Mostafa', 28, 'Finland', 'Espoo', 'female')
print(s1.person_info())
s1.add_skill('JavaScript')
s1.add_skill('React')
s1.add_skill('Python')
print(s1.skills)

print(s2.person_info())
s2.add_skill('Organizing')
s2.add_skill('Marketing')
s2.add_skill('Digital Marketing')
print(s2.skills)

```

1 - We can use super() built-in function or the parent name Person to automatically inherit the methods and properties from its parent. 

2- In the example above we override the parent method. The child method has a different feature, it can identify, 
if the gender is male or female and assign the proper pronoun(He/She).

```python
import numpy as np

# Create a full rank matrix
full_rank_matrix = np.array([[1, 2], [3, 4]])
rank_full = np.linalg.matrix_rank(full_rank_matrix)
print(f"The rank of the full rank matrix is: {rank_full}")
# Output: The rank of the full rank matrix is: 2

# Create a rank-deficient matrix (rows/columns are linearly dependent)
rank_deficient_matrix = np.array([[1, 2], [2, 4]])
rank_deficient = np.linalg.matrix_rank(rank_deficient_matrix)
print(f"The rank of the rank-deficient matrix is: {rank_deficient}")
# Output: The rank of the rank-deficient matrix is: 1
```
